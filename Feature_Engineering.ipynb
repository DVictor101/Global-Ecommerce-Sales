```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
%matplotlib inline
```

```python
f_data= pd.read_csv("./data_folder/cleaned_data.csv")
```

```python
f_data.head(3)
```

## Date Feature Engineering


```python
f_data["Order_Date"] = pd.to_datetime(f_data["Order_Date"])
```

```python
f_data["Order_Date"].dtype
```

```python
# assigning the month data
f_data["Order_Month"] = f_data["Order_Date"].dt.month
f_data["Order_Month"].head(2)
```

```python
## creating the order quarter

f_data["Order_Quarter"] = f_data["Order_Date"].dt.quarter
f_data["Order_Quarter"].head(2)
```

```python
# creating the order year

f_data["Order_Year"] = f_data["Order_Date"].dt.year
f_data["Order_Year"].head(2)
```

## Financial Feature Engineering

```python
## Profit Margin

f_data[["Profit_Margin_%", "Profit", "Revenue"]].head(3)
```

```python
#revenue / unit

f_data["revenue_per_unit"] = f_data["Revenue"] / f_data["Quantity"]
f_data["revenue_per_unit"].head(2)
```

```python
#profit / unit

f_data["profit_per_unit"] = f_data["Profit"] / f_data["Quantity"]
f_data["profit_per_unit"].head(2)
```

```python
# shipping cost to revenue ratio

f_data["scost_per_revenue"] = f_data["Shipping_Cost"] / f_data["Revenue"]
f_data["scost_per_revenue"].head(2)
```

```python
# Net revenue

f_data["net_revenue"] = f_data["Revenue"] / f_data["Shipping_Cost"]
f_data["net_revenue"].head(2)
```
