## Country Profitability Analysis for Global E-Commerce Expansion

### Business Problem

A Chinese e-commerce company plans to expand its operations into a new Country. However, launching in every country is expensive and risky. The company needs a data-driven approach to identify which country is the most profitable and offers the greatest business potential.

This project analyzes historical e-commerce sales data from 2021 to 2024 to determine which country generates the highest revenue and profit. The insights will help the business stakeholders make informed decisions about market expansion and resource allocation.

## Project Objective

The primary objective of this project is to identify the countries that would be the most profitable for launching or expanding an e-commerce business.

To achieve this, the project will analyze sales performance across countries and answer key business questions using exploratory data analysis, statistical methods, and machine learning.


## Business Questions

This project aims to answer the following questions:

1. Which country generates the highest total revenue?
2. Which country generates the highest profit?
3. Which country have the highest average order value?
4. Which product categories perform best in each country?
5. Which regions contribute the largest share of global sales?
6. How have sales changed between 2021 and 2024?
7. Which country has the greatest long-term business potential?
8. What recommendations can be made for future market expansion?
   

## Dataset

**Dataset Name**

Global E-Commerce Sales Dataset (Synthetic) | 2021–2024

**Description**

This synthetic dataset simulates realistic e-commerce transactions from a fictional global online retailer operating in 20 countries across four major regions between January 2021 and December 2024.

The dataset contains information about:

* Order_ID', 'Order_Date', 'Quarter', 'Season', 'Customer_ID',
* 'Customer_Gender', 'Customer_Segment', 'Region', 'Country',
* 'Category', 'Sub_Category', 'Product_Name', 'Shipping_Method',
* 'Payment_Method', 'Order_Status', Year', 'Month', 'Unit_Price',
* 'Quantity', 'Discount', 'Revenue', 'Cost', 'Profit',
* 'Profit_Margin_%', 'Shipping_Cost', 'Shipping_Days'
    
The data is from kaggle Global E-Commerce Sales Dataset (Synthetic) | 2021–2024 https://www.kaggle.com/datasets/abdelfattahibrahim/global-e-commerce-sales-dataset-20212024/data
